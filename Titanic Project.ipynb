## Detect interquartile

```python
def detect_iqr_outliers(dataframe, column):
    Q1 = dataframe[column].quantile(0.25)
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = dataframe[(dataframe[column] < lower_bound) | (dataframe[column] > upper_bound)]
    return outliers
```

```python
age_outliers_iqr = detect_iqr_outliers(titanic_df, 'Age')

print("Age Outliers based on IQR:")
age_outliers_iqr
```

```python
age_outliers_iqr.shape
```

```python
fare_outliers_iqr = detect_iqr_outliers(titanic_df, 'Fare')

print("\nFare Outliers based on IQR:")
fare_outliers_iqr
```

### Apply std for the same columns

```python
def detect_std_outliers(dataframe, column, z_threshold=3):
    mean = dataframe[column].mean()
    std = dataframe[column].std()
    
    outliers = dataframe[(dataframe[column] < mean - z_threshold * std) |
                         (dataframe[column] > mean + z_threshold * std)]
    return outliers
```

```python
age_outliers_std = detect_std_outliers(titanic_df, 'Age')

print("Age Outliers based on Standard Deviation:")
age_outliers_std

```

```python
fare_outliers_std = detect_std_outliers(titanic_df, 'Fare')

print("\nFare Outliers based on Standard Deviation:")
fare_outliers_std
```

Next Step:

After identifying outliers, you need to decide how to handle them based on your analysis or modeling needs. 

Options include:
1. Removing outliers: This can help if the outliers are due to errors or are extreme enough to skew your analysis.
2. Capping values: Using winsorization, where extreme values are replaced with a specified percentile of the data.
3. Keeping them: If they are genuine values and explainable, they might be critical to your analysis, especially in tasks like fraud detection.

```python
# Winsorizing

from scipy.stats.mstats import winsorize

# Winsorize 'Age' and 'Fare' at the 5th and 95th percentiles
titanic_df['Winsorized_Age'] = winsorize(titanic_df['Age'], limits=[0.05, 0.05])
titanic_df['Winsorized_Fare'] = winsorize(titanic_df['Fare'], limits=[0.05, 0.05])
```

# Correlations

provide insights into relationships between different numerical features. Correlation coefficients can indicate whether changes in one variable are associated with changes in another.

```python
titanic_df.head()
```

```python
# Simplify the dataset to relevant features
titanic_df['Sex'] = titanic_df['Sex'].map({'male': 0, 'female': 1})
```

```python
# Drop non-numerical columns for simplicity
numerical_df = titanic_df.select_dtypes(include=['float64', 'int64'])
numerical_df.head()
```

```python
numerical_df.isnull().sum()
```

```python
# Handling missing values by imputing or dropping
numerical_df = numerical_df.fillna(numerical_df.median())
numerical_df.head()
```

```python
numerical_df.isnull().sum()
```

```python
# Calculate the correlation matrix
correlation_matrix = numerical_df.corr()

# Display the correlation matrix
correlation_matrix
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Heatmap of the correlation matrix
plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap='coolwarm', cbar=True)
plt.title('Correlation Matrix of Titanic Dataset')
plt.show()
```

Interpretation
- Positive Correlation: If the coefficient is close to +1, it implies that an increase in one variable is associated with an increase in the other.
- Negative Correlation: If the coefficient is close to -1, it implies that an increase in one variable is associated with a decrease in the other.
- No Correlation: A coefficient close to 0 implies no linear relationship between the variables.

### Apply the Threshold
filter this matrix to find pairs of variables where the correlation is either greater than 0.5 or less than -0.5. It's important to exclude the diagonal values (which are always 1)

```python
high_corr = correlation_matrix[(correlation_matrix > 0.4) | (correlation_matrix < -0.4)]
high_corr
```

### Mask Self-Correlations
Since the correlation of a variable with itself is always 1, these need to be ignored (we can set them to NaN manually or use np.nan on the diagonal).

```python
# Mask the diagonal since we are not interested in self-correlations
np.fill_diagonal(high_corr.values, np.nan)
```

```python
high_corr
```

### Extract the highest corr

```python
# Find and display the significant correlations
significant_pairs = high_corr.unstack().dropna()  # Unstack to pair correlations, drop NaN values

# Filter to include only correlations above 0.5 or below -0.5
significant_pairs = significant_pairs[(significant_pairs > 0.4) | (significant_pairs < -0.4)].drop_duplicates()

print("Pairs with High Correlation:")
significant_pairs
```
