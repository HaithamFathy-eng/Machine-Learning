```python
df = pd.get_dummies(df, columns=['3D_available', 'Genre'], drop_first=True, dtype='uint8')
df.head()
```

```python
X = df.drop(['Start_Tech_Oscar'], axis=1) 
y = df['Start_Tech_Oscar']
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42) 
```

```python
from sklearn.preprocessing import StandardScaler
cat_features = ['3D_available_YES',	'Genre_Comedy',	'Genre_Drama',	'Genre_Thriller']
num_features = [col for col in X_train.columns if col not in cat_features]
num_features
```

```python
from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_features),
        ('cat', 'passthrough', cat_features)
    ]
)

X_train_scaled = preprocessor.fit_transform(X_train)
X_test_scaled = preprocessor.transform(X_test)
```

```python
from sklearn.neighbors import KNeighborsClassifier
knn = KNeighborsClassifier(n_neighbors=7)
knn.fit(X_train, y_train)
```

```python
Knn_Predictions = knn.predict(X_test)
from sklearn.metrics import accuracy_score

knn_accuracy = accuracy_score(y_test, Knn_Predictions)
print(f"Accuracy: {knn_accuracy:.2f}")
```

```python
from sklearn.linear_model import LogisticRegression
LR = LogisticRegression()
LR.fit(X_train, y_train)
```

```python
LR_pred = LR.predict(X_test)
LR_accuracy = accuracy_score(y_test, LR_pred)
print(f"Accuracy: {LR_accuracy:.2f}")
```

```python
from sklearn.svm import SVC

svm = SVC(kernel='linear', probability=True)
svm.fit(X_train, y_train)
```

```python
svm_pred = svm.predict(X_test)
```

```python
svm_accuracy = accuracy_score(y_test, svm_pred)
print(f"Accuracy: {svm_accuracy:.2f}")
```

```python
from sklearn.naive_bayes import GaussianNB

GNB = GaussianNB()
GNB.fit(X_train, y_train)
```

```python
GNB_pred = GNB.predict(X_test)
```

```python
GNB_accuracy = accuracy_score(y_test, GNB_pred)
print(f"Accuracy: {GNB_accuracy:.2f}")
```

```python
from sklearn.tree import DecisionTreeClassifier
clf=DecisionTreeClassifier(criterion='entropy',max_depth=3)
clf.fit(X_train,y_train)
```

```python
y_pred_tree=clf.predict(X_test)
```

```python
from sklearn import metrics
print("Accuracy :",metrics.accuracy_score(y_test,y_pred_tree))
```

```python
from sklearn.model_selection import  GridSearchCV
parameters={'max_depth':range(3,10)}
clf_ = GridSearchCV(DecisionTreeClassifier(criterion='gini'), parameters)
clf_.fit(X_train,y_train)
tree_model = clf_.best_estimator_
print (clf_.best_score_, clf_.best_params_)
```
